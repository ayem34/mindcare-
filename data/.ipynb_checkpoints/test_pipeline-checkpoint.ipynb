```python
# Cellule 5 : Tester une prédiction globale
from backend.models.fusion.predict_global import predict_global_from_raw

# Exemple d’entrée : 7 entrées comportementales (jour 1 à jour 7)
# Ici on simule avec 7 dicts identiques (en vrai ils devraient varier)
daily_entries = [
    {
        "heures_sommeil": 6.5,
        "stress_level": 7.0,
        "anxiety_level": 6.0,
        "social_media_hours": 4.5,
        "physical_activity": 1.0,
        "family_history": 0.0,
        "coping_struggles": 1.0,
        "mood_swings": 5.0,
        "days_indoors": 4.0,
    }
] * 7   # 7 jours identiques

# Exemple de 7 textes (un par jour)
daily_texts = [
    "Aujourd'hui je me sens très fatigué et sans motivation.",
    "J'ai du mal à dormir, je me réveille plusieurs fois par nuit.",
    "Je me sens inutile, rien ne me plaît.",
    "Les pensées négatives reviennent sans cesse.",
    "Je n'arrive pas à me concentrer au travail.",
    "Je pleure sans raison apparente.",
    "Je me sens vide et isolé."
]

result = predict_global_from_raw(daily_entries, daily_texts)

print("\n=== RÉSULTAT FUSION ===")
print(f"Score final      : {result['score_final']:.4f}")
print(f"Niveau de risque : {result['niveau_risque']}")
print(f"Message          : {result['message_fr']}")
print(f"Signal suicidaire: {result['signal_suicidaire']}")
print("\nDétails :")
for k, v in result['details'].items():
    print(f"  {k}: {v}")
```

```python
import pandas as pd
from backend.utils.config import CSV_TEEN

df_teen = pd.read_csv(CSV_TEEN)
depressif = df_teen[df_teen['depression_label'] == 1].iloc[0]

# Afficher ses caractéristiques
print("=== Adolescent dépressif ===")
print(f"Âge: {depressif['age']}")
print(f"Sommeil: {depressif['sleep_hours']} h")
print(f"Stress: {depressif['stress_level']}")
print(f"Anxiété: {depressif['anxiety_level']}")
print(f"Réseaux sociaux: {depressif['daily_social_media_hours']} h")
print(f"Activité physique: {depressif['physical_activity']}")
```

```python
#2. Extraire un adolescent sain
```

```python
sain = df_teen[df_teen['depression_label'] == 0].iloc[0]
print("\n=== Adolescent sain ===")
print(f"Âge: {sain['age']}")
print(f"Sommeil: {sain['sleep_hours']} h")
print(f"Stress: {sain['stress_level']}")
print(f"Anxiété: {sain['anxiety_level']}")
print(f"Réseaux sociaux: {sain['daily_social_media_hours']} h")
```

```python
#3. Construire les entrées comportementales au format attendu

def teen_to_entry(row):
    return {
        "heures_sommeil": row['sleep_hours'],
        "stress_level": row['stress_level'],
        "anxiety_level": row['anxiety_level'],
        "social_media_hours": row['daily_social_media_hours'],
        "physical_activity": row['physical_activity'],
        "family_history": 0.0,   # non dispo dans Teen
        "coping_struggles": 0.0,
        "mood_swings": 0.0,
        "days_indoors": 0.0,
    }

entry_dep = teen_to_entry(depressif)
entry_sain = teen_to_entry(sain)

# Répéter sur 7 jours (mêmes valeurs)
daily_entries_dep = [entry_dep] * 7
daily_entries_sain = [entry_sain] * 7
```
